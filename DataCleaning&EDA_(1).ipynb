```python
df1=df.copy()
```

```python
temp_df=df1[df1['price']=='Free']
df1.loc[temp_df.index,'price']=0
df1['price']=df1['price'].astype(float)
df1.info()
```

```python
df1['author']=df1['author'].str.strip().str.split(':').str.get(1)
```

```python
df1['narrator']=df1['narrator'].str.strip().str.split(':').str.get(1)
```

```python
df1['language']=df1['language'].str.lower()
```

```python
df1['stars'].value_counts()
```

```python
def star_rating(text):
  if text=='Not rated yet':
    return '0:0'
  else:
    return text.replace('out of 5 stars',':').replace('ratings','').replace('rating','')
```

```python
df1['star']=df1['stars'].apply(star_rating).str.strip().str.split(':').str.get(0)
df1['ratings']=df1['stars'].apply(star_rating).str.strip().str.split(':').str.get(1)
```

```python
df1.head()
```

```python
df1['star']=df1['star'].astype(float)
df1['ratings']=df1['ratings'].str.replace(',','').astype(float)
```

```python
df1.drop('stars',axis=1,inplace=True)
```

```python
df1['time']=df1['time'].str.strip().str.replace('hrs','h').str.replace('hr','h').str.replace('mins','m').str.replace('min','m')
```

```python
df1
```

```python
temp_df1=df1[~df1['time'].str.contains('and')]
temp_df2=df1[df1['time'].str.contains('and')]
```

```python
temp_df2
```

```python
def time_conv(text):
  if 'Less' in text:
    return 30
  elif 'h'in text:
    hr=text.split(' ')
    return int(hr[0])*60
  elif 'm' in text:
    min=text.split(' ')
    return int(min[0])
  else:
    return int(text)

```

```python
t1=temp_df2['time'].str.strip().str.split('and').str.get(0).apply(time_conv)
```

```python
t2=temp_df2['time'].str.strip().str.split('and').str.get(1)
t2=t2.str.strip().apply(time_conv)
```

```python
temp_df2['time']=t1+t2
```

```python
temp_df2['time']
```

```python
temp_df1['time']=temp_df1['time'].str.strip().apply(time_conv)
```

```python
temp_df1['time']
```

```python
df1.loc[temp_df1.index,'time']=temp_df1['time']
df1.loc[temp_df2.index,'time']=temp_df2['time']
```

```python
df1['time']=df1['time'].astype(int)
```

```python
df1['releasedate']=pd.to_datetime(df1['releasedate'])
```
